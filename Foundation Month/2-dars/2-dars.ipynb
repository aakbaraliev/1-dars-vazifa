# ikkinchi va to'rtinchi vazifalar

```python
a = int(input("Birinchi sonni kiriting: "))
b = int(input("Ikkinchi sonni kiriting: "))

print(a)
print(b)

if a > 0 and a % 2 == 0:
    print("a soni musbat va juft")
elif a > 0 and a % 2 != 0:
    print("a soni musbat va toq")
elif a < 0 and a % 2 == 0:
    print("a soni manfiy va juft")
elif a < 0 and a % 2 != 0:
    print("a soni manfiy va toq")
else:
    print("nol")

if b > 0 and b % 2 == 0:
    print("b soni musbat va juft")
elif b > 0 and b % 2 != 0:
    print("b soni musbat va toq")
elif b < 0 and b % 2 == 0:
    print("b soni manfiy va juft")
elif b < 0 and b % 2 != 0:
    print("b soni manfiy va toq")
else:
    print("nol")
```
